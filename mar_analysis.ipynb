```python
features = ['Duration', 'Impressions', 'Clicks', 'Leads',
    'Conversions', 'Acquisition_Cost', 'Engagement_Score']
valid = df['Profit_Status'].notna()

x= df.loc[valid, features].copy()
y = df.loc[valid, 'Profit_Status'].copy()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer

x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.3, random_state=42, stratify=y)

imputer= SimpleImputer(strategy='median')

x_train = imputer.fit_transform(x_train)
x_test = imputer.transform(x_test)
```

Try different classification methods
1. Decision Tree Classifier
2. Random Forest Classifier
3. Logistic Regression

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (accuracy_score, classification_report, confusion_matrix)

model_decision_tree = DecisionTreeClassifier(max_depth=15, random_state=42)
model_decision_tree.fit(x_train, y_train)
y_prediction = model_decision_tree.predict(x_test)

print('Accuracy:', accuracy_score(y_test, y_prediction))
print('Classification Report:', classification_report(y_test, y_prediction, zero_division=0))
print('Confusion Matrix:', confusion_matrix(y_test, y_prediction))
```

```python
from sklearn.ensemble import RandomForestClassifier

model_random = RandomForestClassifier(n_estimators=100, max_depth=12, min_samples_leaf=5, n_jobs = -1)
model_random.fit(x_train, y_train)
y_pred = model_random.predict(x_test)

print('Accuracy:', accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred, zero_division =0))
print('Confusuin Matrix:', confusion_matrix(y_test, y_pred))
```

```python
from sklearn.linear_model import LogisticRegression

model_logistic = LogisticRegression(max_iter = 1000)
model_logistic.fit(x_train, y_train)
y_predic = model_logistic.predict(x_test)

print('Accuracy:', accuracy_score(y_test, y_predic))
print('Classification Report:', classification_report(y_test, y_predic, zero_division = 0))
print('Confusion Matrix:', confusion_matrix(y_test, y_predic))
```
